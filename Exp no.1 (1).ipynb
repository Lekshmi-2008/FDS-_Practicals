```python
import pandas as pd
import matplotlib.pyplot as plt
data ={
    "Year":[2011,2012,2013,2014,2015,2016,2017,2018,2019,2020],
    "Job_Postings":[180,320,470,650,850,1200,1650,2200,3000,4000]
}
df=pd.DataFrame(data)
print(df)
plt.plot(df["Year"], df["Job_Postings"], marker="o")
plt.title("Trend of Data Science Job Postigs")
plt.xlabel("Year")
plt.ylabel("Number of Job Postings") 
plt.show()
```

```python
## 1)b. Distribution of data Science roles
```

```python
import pandas as pd
import matplotlib.pyplot as plt

roles = {
    "Job_Role": [
        "Data Scientist",
        "Data Analyst",
        "Data Engineer",
        "Machine Learning Engineer",
        "Data Scientist",
        "Data Analyst",
        "Data Engineer",
        "Data Scientist",
        "Data Analyst",
        "Machine Learning Engineer",
        "Data Engineer",
        "Data Scientist"
    ]
}

df = pd.DataFrame(roles)

role_distribution = df["Job_Role"].value_counts()

print(role_distribution)

role_distribution.plot(kind="bar")

plt.title("Distribution of Data Science Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Number of Job Postings")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
## 1)c. Structured.SemiStructured and UnStructured Data
```

```python
import pandas as pd
import json

structured_data = {
    "ID": [101, 102, 103],
    "Name": ["Arun", "Priya", "Karthik"],
    "Department": ["CSE", "IT", "CSE"]
}

structured_df = pd.DataFrame(structured_data)

print("STRUCTURED DATA")
print(structured_df)

semi_structured_data = {
    "students": [
        {"ID": 101, "Name": "Arun", "Department": "CSE"},
        {"ID": 102, "Name": "Priya", "Department": "IT"},
        {"ID": 103, "Name": "Karthik", "Department": "CSE"}
    ]
}

print("\nSEMI-STRUCTURED DATA")
print(json.dumps(semi_structured_data, indent=4))

unstructured_data = """
Arun is a Computer Science student.
Priya is an Information Technology student.
Karthik is interested in Data Science.
"""

print("\nUNSTRUCTURED DATA")
print(unstructured_data)
```

```python
## 1)d. Encrytion and Decrytipon using Fernet
```

```python
from cryptography.fernet import Fernet

key = Fernet.generate_key()

f = Fernet(key)

data = "My secret data"

encrypted = f.encrypt(data.encode())

decrypted = f.decrypt(encrypted).decode()

print("Original Data:", data)
print("Encrypted Data:", encrypted)
print("Decrypted Data:", decrypted)
```
